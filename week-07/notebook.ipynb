```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("retail_store_sales.csv")
data
```

```python
data.info()
```

```python
data.columns = (
    data.columns
      .str.lower()
      .str.strip()
      .str.replace(r"\s+", "_", regex=True)
)

data
```

```python
unique_combinations = (
    data[["item", "price_per_unit"]]
    .dropna(subset=["item"])   # item NaN olanları at
    .drop_duplicates()
    .values.tolist()
)
item_price_map = dict(unique_combinations)
item_price_map
```

```python
data["price_per_unit"] = data["total_spent"] / data["quantity"]
data.info()
```

```python
price_item_map = {v: k for k, v in item_price_map.items()}

mask = data["item"].isna()

data.loc[mask, "item"] = (
    data.loc[mask, "price_per_unit"]
    .map(price_item_map)
)

data.info()
```

```python
data = data.drop_duplicates()
data.info()
```
